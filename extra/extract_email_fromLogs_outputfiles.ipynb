```python
import os
import pandas as pd
import pickle

# Folder containing CSV files
folder_path = r"E:\EmailTool-V2\Barzaan\email_creator_app\files\merger"  # 🔁 Replace with your folder path

# Dictionary to hold Company → Email mapping
company_email_dict = {}

# Loop through all CSV files in the folder
for filename in os.listdir(folder_path):
    if filename.endswith(".csv"):
        file_path = os.path.join(folder_path, filename)
        try:
            df = pd.read_csv(file_path)
            
            # Filter only rows with both Company and Email present
            df = df.dropna(subset=['Company', 'Email'])

            # Add to dictionary (latest file will overwrite duplicates)
            for company, email in zip(df['Company'], df['Email']):
                company_email_dict[company.strip()] = email.strip()
        
        except Exception as e:
            print(f"❌ Failed to read {filename}: {e}")

# Save to a .pkl file
with open("merged_company_email.pkl", "wb") as f:
    pickle.dump(company_email_dict, f)

print("✅ Pickle file 'merged_company_email.pkl' created successfully.")
```

```python
import pandas as pd
import pickle

# Path to your CSV file
csv_file = r'E:\EmailTool-V2\Barzaan\email_creator_app\files\main_database\old_database.csv'  # 🔁 Replace with your actual CSV file name

# Load the CSV
df = pd.read_csv(csv_file)
# Key	Value
# Drop rows where Company or Email is missing
df = df.dropna(subset=["Key", "Value"])

# Build dict without duplicates (lowercase keys for consistency)
company_email_dict = {}
for company, email in zip(df["Key"], df["Value"]):
    company = str(company).strip().lower()
    email = str(email).strip()
    if company and company not in company_email_dict:
        company_email_dict[company] = email

# Save to pickle file
with open("new_database_backup.pkl", "wb") as f:
    pickle.dump(company_email_dict, f)

print("✅ Pickle file 'company_email.pkl' created successfully.")
```

```python
import pickle
import os

def safe_write_pickle(data, filepath):
    temp_path = filepath + ".tmp"
    with open(temp_path, "wb") as f:
        pickle.dump(data, f)
    os.replace(temp_path, filepath)  # atomic on most OS
safe_write_pickle(email_patterns, "tempbackup.pkl")
```
