```python
import pickle
import os

old_db_path = r"E:\EmailTool-V2\Barzaan\email_creator_app\APP\files\main_database\old_db.pkl"
new_db_path = r"E:\EmailTool-V2\Barzaan\email_creator_app\APP\files\main_database\new_db.pkl"

# -------------------------------
# Load OLD DB
# -------------------------------
with open(old_db_path, "rb") as f:
    old_db = pickle.load(f)

# -------------------------------
# Load NEW DB
# -------------------------------
with open(new_db_path, "rb") as f:
    new_db = pickle.load(f)

print(f"Old DB size before update: {len(old_db)}")
print(f"New DB size: {len(new_db)}")

# -------------------------------
# Merge logic
# -------------------------------
added = 0
updated = 0

for company, email_pattern in new_db.items():
    if company in old_db:
        if old_db[company] != email_pattern:
            old_db[company] = email_pattern
            updated += 1
    else:
        old_db[company] = email_pattern
        added += 1

print(f"Added new companies: {added}")
print(f"Updated existing companies: {updated}")
print(f"Old DB size after update: {len(old_db)}")

# -------------------------------
# Backup OLD DB (IMPORTANT)
# -------------------------------
backup_path = old_db_path + ".backup"
if not os.path.exists(backup_path):
    os.rename(old_db_path, backup_path)
    print(f"Backup created: {backup_path}")

# -------------------------------
# Save updated OLD DB
# -------------------------------
with open(old_db_path, "wb") as f:
    pickle.dump(old_db, f)

print("✅ Old database successfully updated with new database")
```

```python
import pickle

new_db_path = r"E:\EmailTool-V2\Barzaan\email_creator_app\Email_Tool-2026\APP\files\main_database\old_db_copy_2026.pkl"

INVALID_VALUES = {"search failed", "unknown pattern","no valid pattern found"}

# -------------------------------
# Load NEW DB
# -------------------------------
with open(new_db_path, "rb") as f:
    data = pickle.load(f)

if not isinstance(data, dict):
    raise TypeError("new_db.pkl must contain a dictionary")

original_count = len(data)

# -------------------------------
# Remove invalid values
# -------------------------------
cleaned_data = {
    k: v
    for k, v in data.items()
    if str(v).strip().lower() not in INVALID_VALUES
}

removed_count = original_count - len(cleaned_data)

# -------------------------------
# Save back to SAME file
# -------------------------------
with open(new_db_path, "wb") as f:
    pickle.dump(cleaned_data, f)

print("✅ new_db.pkl cleaned successfully")
print(f"🔑 Original entries : {original_count}")
print(f"❌ Removed entries  : {removed_count}")
print(f"📦 Remaining entries: {len(cleaned_data)}")
```
